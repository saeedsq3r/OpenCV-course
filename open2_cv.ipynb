# **class-30**
## Saving a Video using OpenCV

```python
import cv2
import numpy as np
```

```python
vd = cv2.VideoCapture(0)

f = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter('picture/demo.mp4',f,40.0,(640,480),0)

while vd.isOpened():
    r,frame = vd.read()
    if r == True:
        frame = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        frame = cv2.flip(frame,1)
        out.write(frame)
        cv2.imshow('video',frame)
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break
    else:
        break

vd.release()
out.release()
cv2.destroyAllWindows()
```

# **class-37**
## Histograms Equalization In OpenCV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

- Normal equalizer

```python
img = cv2.imread('picture/equa.jpg')
img = cv2.resize(img,(500,500))

gry = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

hist = cv2.calcHist([gry],[0],None,[255],[0,255])
eq = cv2.equalizeHist(gry)
hist1 = cv2.calcHist([eq],[0],None,[255],[0,255])


plt.plot(hist1,c = 'r')

plt.plot(hist)
plt.show()



cv2.imshow('new eq',eq)
cv2.imshow('equalizer',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread('picture/equa1.jpg')
img = cv2.resize(img,(500,500))

gry = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

hist = cv2.calcHist([gry],[0],None,[255],[0,255])
eq = cv2.equalizeHist(gry)
hist1 = cv2.calcHist([eq],[0],None,[255],[0,255])


cl = cv2.createCLAHE(clipLimit=1,tileGridSize=(8,8))
f = cl.apply(gry)





plt.plot(hist1,c = 'r')
plt.plot(f,c = 'g')
plt.plot(hist)
plt.show()



cv2.imshow('new clahe',f)
cv2.imshow('new eq',eq)
cv2.imshow('equalizer',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# **class-40**
## Contour Moments and Convex Hull using OpenCV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('picture/cont.png')

gry = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

r,th = cv2.threshold(gry,220,255,cv2.THRESH_BINARY)

cnt,hir = cv2.findContours(th,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)

ar = []
for c in cnt:
    m = cv2.moments(c)
    x = int(m["m10"]/m['m00'])
    y = int(m["m01"]/m['m00'])
    cv2.drawContours(img,cnt,-1,(0,0,255),4)
    cv2.circle(img,(x,y),5,(255,0,0),-1)
    a = cv2.contourArea(c)
    ar.append(a)

    ep = 0.01*cv2.arcLength(c,True)
    d = cv2.approxPolyDP(c,ep,True)
    h = cv2.convexHull(d)
    x,y,w,h = cv2.boundingRect(h)
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),4)



cv2.imshow('img_th',th)
# cv2.imshow('img_gry',gry)
cv2.imshow('img',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
