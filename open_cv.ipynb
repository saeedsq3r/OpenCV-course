# **class-9**
## Bitwise Operation on Images using OpenCV (Bitwise Operation on Binary Images)


```python

import cv2
import numpy as np
```

```python
img1 = cv2.imread('picture/img11.jpeg')
img2 = cv2.imread('picture/img13.jpg')

img1 = cv2.resize(img1,(300,300))
img2 = cv2.resize(img2,(300,300))

new = cv2.bitwise_and(img1,img2)
# new = cv2.bitwise_or(img1,img2)
# new = cv2.bitwise_xor(img1,img2)
# new = cv2.bitwise_not(img1)
# new1 = cv2.bitwise_not(img2)

h = np.hstack((img1,new))
h1 = np.hstack((img2,new))

v = np.vstack((h,h1))

cv2.imshow('and',v)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img1 = cv2.imread('picture/img9.jpeg')
img2 = cv2.imread('picture/img8.jpeg')


img1 = cv2.resize(img1,(300,300))
img2 = cv2.resize(img2,(300,300))



new = cv2.bitwise_and(img1,img2)

h = np.hstack((img1,img2,new))

cv2.imshow('and',h)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# **class-11**
## Image Scalling Rotating using OpenCV

```python
import cv2
import numpy as np
```

```python
or_img = cv2.imread('picture/img9.jpeg')

res_img = cv2.resize(or_img,(500,500))


cv2.imshow('saeed1',or_img)
cv2.imshow('saeed',res_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
or_img = cv2.imread('picture/img9.jpeg')

res_img = cv2.resize(or_img,(500,500))

w,h = res_img.shape[0],res_img.shape[1]



m = cv2.getRotationMatrix2D((w/2,h/2),90,1)
new_img = cv2.warpAffine(res_img,m,(h,w))


h = np.hstack([res_img,new_img])

cv2.imshow('saeed',h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
or_img = cv2.imread('picture/img9.jpeg')
res_img = cv2.resize(or_img,(500,500))

w,h = res_img.shape[0] , res_img.shape[1]



m = cv2.getRotationMatrix2D((w/2,h/2),180,1)
new_img = cv2.warpAffine(res_img,m,(h,w))

h = np.hstack((res_img,new_img))

cv2.imshow('saeed',h)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# **class-15**
## PLAY AIDEO USING OPENCV

```python
import cv2
import numpy as np
```

```python
cap = cv2.VideoCapture('picture/video1.mp4')

while cap.isOpened():
    r,frame = cap.read()
    if r == True:
        frame = cv2.resize(frame,(600,500))
        cv2.imshow('video_play',frame)
        if cv2.waitKey(25) & 0xff == ord('p'):
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```

# **class-18**
## Morphological Operations using OpenCV - 1

```python
import cv2
import numpy as np
```

```python
img = cv2.imread('picture/j.png')
re_img = cv2.resize(img,(500,500))

m = np.ones((15,15),np.int8)
# print(m)
er = cv2.erode(re_img,m,iterations=1)
di = cv2.dilate(re_img,m,iterations=1)

cv2.imshow('morphological',re_img)  
cv2.imshow('morphological_erode',er)
cv2.imshow('morphological_dilate',di)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
